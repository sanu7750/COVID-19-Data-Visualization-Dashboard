## 📌 Project Objectives
1. **Data Ingestion & Cleaning:** Extract and process time-series COVID-19 data across 60+ countries (2020–2023).
2. **Epidemiological Feature Engineering:** Compute 7-day moving averages, normalized per-capita metrics (Cases/Deaths per Million), Case Fatality Rate (CFR %), and Recovery Rate (%).
3. **Country-Wise Comparative Analysis:** Benchmark cumulative caseloads, infection trajectories, and mortality across global regions.
4. **Vaccination Decoupling Analysis:** Statistically examine the relationship between mass immunization campaigns and the reduction in severe disease and mortality.
5. **Tableau Integration:** Prepare structured, clean datasets for interactive Tableau dashboards.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visual Styling Configuration
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.titleweight'] = 'bold'

# Paths
BASE_DIR = os.path.dirname(os.path.abspath(''))
DATA_DIR = os.path.join(BASE_DIR, 'data', 'processed')

print(f"Data directory: {DATA_DIR}")
```

## 1. Load Cleaned Datasets
We load both the daily time-series dataset and the country-level summary snapshot.

```python
ts_path = os.path.join(DATA_DIR, 'covid_19_cleaned_timeseries.csv')
country_path = os.path.join(DATA_DIR, 'covid_19_country_summary.csv')
global_path = os.path.join(DATA_DIR, 'covid_19_global_summary.csv')

df_ts = pd.read_csv(ts_path)
df_ts['date'] = pd.to_datetime(df_ts['date'])

df_country = pd.read_csv(country_path)
df_global = pd.read_csv(global_path)
df_global['date'] = pd.to_datetime(df_global['date'])

print(f"Master Timeseries Shape: {df_ts.shape}")
print(f"Country Summary Shape:   {df_country.shape}")
print(f"Global Summary Shape:    {df_global.shape}")
df_country.head()
```

## 2. Global Headline Figures & Distribution

```python
total_cases = df_country['cumulative_cases'].sum()
total_deaths = df_country['cumulative_deaths'].sum()
total_recovered = df_country['cumulative_recovered'].sum()
total_vax = df_country['total_vaccinations'].sum()
global_cfr = (total_deaths / total_cases) * 100
global_rec_rate = (total_recovered / total_cases) * 100

print(f"=======================================================")
print(f"       GLOBAL COVID-19 HEADLINE EPIDEMIOLOGICAL KPIs")
print(f"=======================================================")
print(f"Total Confirmed Cases:       {total_cases:,.0f}")
print(f"Total Confirmed Fatalities:  {total_deaths:,.0f}")
print(f"Estimated Recoveries:        {total_recovered:,.0f}")
print(f"Total Vaccine Doses Given:   {total_vax:,.0f}")
print(f"Overall Case Fatality Rate:  {global_cfr:.2f}%")
print(f"Overall Recovery Rate:       {global_rec_rate:.2f}%")
print(f"=======================================================")
```

## 3. Global Epidemic Curves & Waves (Cases vs Deaths)
Visualizing the progression of pandemic waves from the initial Wild-Type outbreak in early 2020 through the Alpha, Delta, and Omicron waves.

```python
fig, ax1 = plt.subplots(figsize=(14, 6))

# Daily Cases on Primary Axis
color_cases = '#1f77b4'
ax1.set_xlabel('Timeline (Date)', fontweight='bold')
ax1.set_ylabel('Daily Incident Cases (7-Day Avg)', color=color_cases, fontweight='bold')
line1 = ax1.plot(df_global['date'], df_global['daily_cases_7d_avg'], color=color_cases, linewidth=2, label='Daily Cases (7D Avg)')
ax1.tick_params(axis='y', labelcolor=color_cases)

# Daily Deaths on Secondary Axis
ax2 = ax1.twinx()
color_deaths = '#d62728'
ax2.set_ylabel('Daily Fatalities (7-Day Avg)', color=color_deaths, fontweight='bold')
line2 = ax2.plot(df_global['date'], df_global['daily_deaths_7d_avg'], color=color_deaths, linewidth=2, linestyle='--', label='Daily Fatalities (7D Avg)')
ax2.tick_params(axis='y', labelcolor=color_deaths)

# Combine legends
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left')

plt.title('Global COVID-19 Trajectory: Daily Cases vs. Daily Fatalities (Dual-Axis)', pad=15)
plt.tight_layout()
plt.show()
```

## 4. Country-Wise Comparative Analysis: Top 10 Impacted Nations
Benchmarking the highest caseloads, fatal impacts, and per-capita normalization.

```python
top_cases = df_country.sort_values(by='cumulative_cases', ascending=False).head(10)
top_deaths = df_country.sort_values(by='cumulative_deaths', ascending=False).head(10)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Top 10 by Total Cases
sns.barplot(ax=axes[0], data=top_cases, x='cumulative_cases', y='country', palette='Blues_r')
axes[0].set_title('Top 10 Countries by Total Confirmed Cases')
axes[0].set_xlabel('Cumulative Cases')
axes[0].set_ylabel('Country')

# Top 10 by Total Deaths
sns.barplot(ax=axes[1], data=top_deaths, x='cumulative_deaths', y='country', palette='Reds_r')
axes[1].set_title('Top 10 Countries by Cumulative Fatalities')
axes[1].set_xlabel('Cumulative Deaths')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

## 5. Normalized Metric Comparison: Cases & Deaths Per Million People
Absolute counts are biased toward populous countries. Normalizing by population provides a true comparative infection and mortality burden.

```python
top_cases_per_mil = df_country.sort_values(by='cases_per_million', ascending=False).head(10)
top_deaths_per_mil = df_country.sort_values(by='deaths_per_million', ascending=False).head(10)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(ax=axes[0], data=top_cases_per_mil, x='cases_per_million', y='country', palette='Oranges_r')
axes[0].set_title('Top 10 Countries: Infection Rate (Cases per Million)')
axes[0].set_xlabel('Cases per Million People')

sns.barplot(ax=axes[1], data=top_deaths_per_mil, x='deaths_per_million', y='country', palette='Purples_r')
axes[1].set_title('Top 10 Countries: Mortality Rate (Deaths per Million)')
axes[1].set_xlabel('Deaths per Million People')

plt.tight_layout()
plt.show()
```

## 6. Vaccination Coverage vs. Case Fatality Rate (CFR %)
Analyzing the clinical efficacy of global immunization campaigns in decoupling infection from mortality.

```python
plt.figure(figsize=(12, 7))
sns.regplot(
    data=df_country,
    x='fully_vaccinated_pct',
    y='case_fatality_rate_pct',
    scatter_kws={'alpha': 0.7, 's': df_country['population'] / 5_000_000, 'color': '#2e75b6'},
    line_kws={'color': '#c00000', 'linewidth': 2}
)

plt.title('Clinical Decoupling: Fully Vaccinated Population (%) vs Case Fatality Rate (%)', pad=15)
plt.xlabel('Fully Vaccinated Population Share (%)', fontweight='bold')
plt.ylabel('Case Fatality Rate (CFR %)', fontweight='bold')

# Annotate selected major countries
for _, row in df_country.sample(min(8, len(df_country)), random_state=42).iterrows():
    plt.annotate(
        row['country'],
        (row['fully_vaccinated_pct'], row['case_fatality_rate_pct']),
        textcoords="offset points",
        xytext=(5, 5),
        fontsize=9,
        fontweight='semibold'
    )

plt.tight_layout()
plt.show()
```

## 7. Key Analytical Takeaways
1. **Wave Dynamics:** The Omicron variant produced the highest transmission surges in pandemic history, yet was accompanied by a historically low Case Fatality Rate.
2. **Vaccine Decoupling:** Countries with higher fully vaccinated population percentages (>75%) experienced significantly lower mortality burdens during the 2021–2022 waves.
3. **Country-Wise Disparities:** Per-capita normalizations reveal that high-testing developed nations often recorded higher per-million case rates, while clinical outcomes heavily correlated with healthcare capacity and immunization speed.
